```python

import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

# Upload the cleaned sales dataset
uploaded = files.upload()
filename = next(iter(uploaded))
df = pd.read_csv(filename)

print("DATA VISUALIZATION DASHBOARD")
print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

# Chart 1: Total sales by product
if {"product", "sales"}.issubset(df.columns):
    sales_product = df.groupby("product")["sales"].sum().sort_values(
        ascending=False
    )
    sales_product.plot(kind="bar", title="Total Sales by Product")
    plt.xlabel("Product")
    plt.ylabel("Total Sales")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# Chart 2: Total sales by region
if {"region", "sales"}.issubset(df.columns):
    sales_region = df.groupby("region")["sales"].sum()
    sales_region.plot(kind="bar", title="Total Sales by Region")
    plt.xlabel("Region")
    plt.ylabel("Total Sales")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# Chart 3: Distribution of sales
if "sales" in df.columns:
    plt.hist(df["sales"].dropna(), bins=5, edgecolor="black")
    plt.title("Distribution of Sales")
    plt.xlabel("Sales")
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()

# Chart 4: Quantity versus sales
if {"quantity", "sales"}.issubset(df.columns):
    plt.scatter(df["quantity"], df["sales"])
    plt.title("Quantity vs Sales")
    plt.xlabel("Quantity")
    plt.ylabel("Sales")
    plt.tight_layout()
    plt.show()

print("Task 3 visualizations completed!")
```
